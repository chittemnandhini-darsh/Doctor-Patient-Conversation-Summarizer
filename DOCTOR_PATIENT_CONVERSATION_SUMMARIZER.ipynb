```python
# ============================================================
# 🩺 DOCTOR-PATIENT CONVERSATION SUMMARIZER
# Text and Speech Analysis Application
# Google Colab - Single Cell
# ============================================================

!pip -q install gradio SpeechRecognition pydub nltk

import gradio as gr
import speech_recognition as sr
from pydub import AudioSegment
import re

# ============================================================
# SPEECH TO TEXT + SUMMARY
# ============================================================

def summarize_conversation(audio_file):

    if audio_file is None:
        return "Please upload an audio file.", ""

    try:

        # ----------------------------------------------------
        # Convert MP3 to WAV
        # ----------------------------------------------------

        if audio_file.lower().endswith(".mp3"):

            audio = AudioSegment.from_mp3(audio_file)

            wav_file = "doctor_patient_conversation.wav"

            audio.export(wav_file, format="wav")

        else:
            wav_file = audio_file

        # ----------------------------------------------------
        # SPEECH RECOGNITION
        # ----------------------------------------------------

        recognizer = sr.Recognizer()

        with sr.AudioFile(wav_file) as source:

            audio_data = recognizer.record(source)

        text = recognizer.recognize_google(audio_data)

        # ----------------------------------------------------
        # TEXT PROCESSING
        # ----------------------------------------------------

        sentences = re.split(r'[.!?]+', text)

        sentences = [
            sentence.strip()
            for sentence in sentences
            if sentence.strip()
        ]

        # ----------------------------------------------------
        # KEYWORD LISTS
        # ----------------------------------------------------

        symptom_words = [
            "headache",
            "fever",
            "cough",
            "cold",
            "pain",
            "tired",
            "fatigue",
            "vomiting",
            "nausea",
            "dizziness",
            "breathing",
            "stomach",
            "chest pain",
            "sore throat"
        ]

        medicine_words = [
            "tablet",
            "medicine",
            "medication",
            "paracetamol",
            "ibuprofen",
            "antibiotic",
            "syrup",
            "capsule"
        ]

        advice_words = [
            "rest",
            "drink",
            "take",
            "avoid",
            "exercise",
            "sleep",
            "water",
            "follow",
            "consult"
        ]

        # ----------------------------------------------------
        # EXTRACT INFORMATION
        # ----------------------------------------------------

        symptoms = []
        medicines = []
        advice = []

        for sentence in sentences:

            lower_sentence = sentence.lower()

            # Symptoms
            for word in symptom_words:

                if word in lower_sentence:
                    if sentence not in symptoms:
                        symptoms.append(sentence)
                    break

            # Medicines
            for word in medicine_words:

                if word in lower_sentence:
                    if sentence not in medicines:
                        medicines.append(sentence)
                    break

            # Advice
            for word in advice_words:

                if word in lower_sentence:
                    if sentence not in advice:
                        advice.append(sentence)
                    break

        # ----------------------------------------------------
        # CREATE SUMMARY
        # ----------------------------------------------------

        summary = ""

        summary += "🩺 DOCTOR-PATIENT CONVERSATION SUMMARY\n\n"

        summary += "📋 Conversation:\n"

        for sentence in sentences:
            summary += "• " + sentence + "\n"

        summary += "\n"

        summary += "🤒 Symptoms Mentioned:\n"

        if symptoms:
            for item in symptoms:
                summary += "• " + item + "\n"
        else:
            summary += "• No symptoms detected\n"

        summary += "\n"

        summary += "💊 Medicines Mentioned:\n"

        if medicines:
            for item in medicines:
                summary += "• " + item + "\n"
        else:
            summary += "• No medicines detected\n"

        summary += "\n"

        summary += "💡 Doctor's Advice / Recommendations:\n"

        if advice:
            for item in advice:
                summary += "• " + item + "\n"
        else:
            summary += "• No advice detected\n"

        summary += "\n"

        summary += "⚠️ Note:\n"
        summary += (
            "This is an automatically generated summary "
            "for educational purposes and is not medical advice."
        )

        return text, summary

    except sr.UnknownValueError:

        return (
            "Sorry, the speech could not be understood.",
            ""
        )

    except sr.RequestError:

        return (
            "Speech recognition service is unavailable.",
            ""
        )

    except Exception as e:

        return (
            "Error: " + str(e),
            ""
        )


# ============================================================
# GRADIO WEB APPLICATION
# ============================================================

with gr.Blocks(
    title="Doctor-Patient Conversation Summarizer"
) as app:

    gr.Markdown(
        """
        # 🩺 Doctor-Patient Conversation Summarizer

        ### Convert a doctor-patient conversation into an organized summary.

        **🎤 Audio → 🗣️ Speech-to-Text → 🔍 Text Analysis → 📋 Summary**
        """
    )

    gr.Markdown(
        """
        ### 📌 Example Conversation

        **Doctor:** What problem are you having?

        **Patient:** I have been having a headache for three days.

        **Doctor:** Do you have a fever?

        **Patient:** No, but I feel tired.

        **Doctor:** Get enough rest and drink plenty of water.
        """
    )

    # --------------------------------------------------------
    # AUDIO INPUT
    # --------------------------------------------------------

    audio_input = gr.Audio(
        sources=["upload"],
        type="filepath",
        label="🎤 Upload Doctor-Patient Conversation"
    )

    # --------------------------------------------------------
    # BUTTON
    # --------------------------------------------------------

    summarize_button = gr.Button(
        "📋 Generate Summary",
        variant="primary"
    )

    # --------------------------------------------------------
    # SPEECH OUTPUT
    # --------------------------------------------------------

    gr.Markdown("## 🗣️ Converted Conversation")

    speech_output = gr.Textbox(
        label="Speech-to-Text",
        lines=8,
        placeholder="Converted conversation will appear here..."
    )

    # --------------------------------------------------------
    # SUMMARY OUTPUT
    # --------------------------------------------------------

    gr.Markdown("## 📄 Conversation Summary")

    summary_output = gr.Textbox(
        label="Summary",
        lines=15,
        placeholder="Generated summary will appear here..."
    )

    # --------------------------------------------------------
    # BUTTON ACTION
    # --------------------------------------------------------

    summarize_button.click(
        fn=summarize_conversation,
        inputs=audio_input,
        outputs=[
            speech_output,
            summary_output
        ]
    )

    # --------------------------------------------------------
    # FEATURES
    # --------------------------------------------------------

    gr.Markdown(
        """
        ---

        ### ✨ Features

        🎤 Audio upload
        🗣️ Speech-to-text conversion
        🔍 Text analysis
        🤒 Symptom detection
        💊 Medicine detection
        💡 Advice detection
        📋 Automatic conversation summary
        🌐 Simple web application

        **Text and Speech Analysis Project**
        """
    )


# ============================================================
# LAUNCH
# ============================================================

app.launch(share=True)
```
